```python
import numpy as np
vx=np.zeros((3,u850.shape[0]-6,u850.shape[1],u850.shape[2]),dtype='float32')
vy=np.zeros((9,u850.shape[0]-6,u850.shape[1],u850.shape[2]),dtype='float32')
vx[0,:,:]=sp[:-6,:,:]
vx[1,:,:]=st[:-6,:,:]
vx[2,:,:]=sd[:-6,:,:]
vy[0,:,:,:]=u850[6:,:,:]
vy[1,:,:,:]=u925[6:,:,:]
vy[2,:,:,:]=u1000[6:,:,:]
vy[3,:,:,:]=v850[6:,:,:]
vy[4,:,:,:]=v925[6:,:,:]
vy[5,:,:,:]=v1000[6:,:,:]
vy[6,:,:,:]=w850[6:,:,:]
vy[7,:,:,:]=w925[6:,:,:]
vy[8,:,:,:]=w1000[6:,:,:]
vx=vx.transpose(1,2,3,0)
vy=vy.transpose(1,2,3,0)
print(vx.shape,vy.shape)
window_offset = 5
test_size, valid_size = 0.2, 0.1
test_start = int((1 - test_size) * (len(vy) - window_offset))
train_stop = int((1 - valid_size / (1 - test_size)) * test_start)
train_inputs = vx[:train_stop + window_offset]
train_labels = vy[window_offset:train_stop + window_offset]
vx_mean = np.nanmean(train_inputs, axis=0)
vx_std = np.nanstd(train_inputs, axis=0)
vy_mean = np.nanmean(train_labels, axis=0)
vy_std = np.nanstd(train_labels, axis=0)
vx = (vx - vx_mean) / vx_std
vy = (vy - vy_mean) / vy_std
del train_inputs, train_labels
```

```python
model,predicty,testy,r,p,heatmap,weights=Auto_DeepLearning_pytorch(vy,vx,[['cov',[16,3,3]],['batchnormalization'],['activation','relu'],['maxpooling',[2,2]],['cov',[32,3,3]],['batchnormalization'],['activation','tanh'],['maxpooling',[2,2]],['flatten'],['lstm'],['fc',32193]],lat=None,lon=None,a_mode='d',r_number=None,u=None,v=None,timestep=6,outputstep=None,dilations=None,ifmulti_scale='no',num_channels=None,test_size=0.2,valid_size=0.1,k_fold=None,task_mode='regression',if_best_mode='no',modelpath=None,ifrandom_split='no',cov_padding='same',cov_strides=1,pooling_strides=2,dropout=0.0,residual_deep=1,residual_activation='relu',residual_norm='batchnormalization',vit_patch_size=None,vit_dim_feedforward=16,vit_deep=1,vit_num_heads=1,swin_deep=[2,2],swin_num_heads=[1,1],cbam_reduction=16,chebnet_k=3,gsage_aggr='mean',gsage_lstm_units=None,transformer_encoder_deep=1,transformer_num_heads=1,transformer_dim_feedforward=128,tcn_kernel_size=2,nlp_units=128,nlp_num_layers=1,ltc_solver='fused',ltc_ode_unfolds=4,ltc_dt=1.0,ltc_tau_min=1e-4,ltc_activation='softplus',ltc_dropout=None,mamba_state=64,mamba_conv=4,mamba_expand=2,activation='relu',if_print_model='yes',loss_function='default',optimizer='Adam',metrics='Pearsonr',learning_rate=0.0001,lr_decay=None,regularization=None,regularization_lambda=1e-5,epochs=5000,batch_size=500,if_early_stopping=100,ifheatmap='no',ifweight='no',ifmute='no',ifsave='yes',savepath=str(MODEL_DIR / 'CNN_LSTM_6_24year_changjiang'),device='gpu')
```
