```python
import numpy as np
vx=np.zeros((3,u850.shape[0]-5,u850.shape[1],u850.shape[2]),dtype='float32')
vy=np.zeros((9,u850.shape[0]-5,u850.shape[1],u850.shape[2]),dtype='float32')
vx[0,:,:]=sp[:-5,:,:]
vx[1,:,:]=st[:-5,:,:]
vx[2,:,:]=sd[:-5,:,:]
vy[0,:,:,:]=u850[5:,:,:]
vy[1,:,:,:]=u925[5:,:,:]
vy[2,:,:,:]=u1000[5:,:,:]
vy[3,:,:,:]=v850[5:,:,:]
vy[4,:,:,:]=v925[5:,:,:]
vy[5,:,:,:]=v1000[5:,:,:]
vy[6,:,:,:]=w850[5:,:,:]
vy[7,:,:,:]=w925[5:,:,:]
vy[8,:,:,:]=w1000[5:,:,:]
vx=vx.transpose(1,2,3,0)
vy=vy.transpose(1,2,3,0)
print(vx.shape,vy.shape)
window_offset = 5
test_size, valid_size = 0.2, 0.1
test_start = int((1 - test_size) * (len(vy) - window_offset))
train_stop = int((1 - valid_size / (1 - test_size)) * test_start)
train_inputs = vx[:train_stop + window_offset]
train_labels = vy[window_offset:train_stop + window_offset]
vx_mean = np.nanmean(train_inputs, axis=0)
vx_std = np.nanstd(train_inputs, axis=0)
vy_mean = np.nanmean(train_labels, axis=0)
vy_std = np.nanstd(train_labels, axis=0)
vx = (vx - vx_mean) / vx_std
vy = (vy - vy_mean) / vy_std
del train_inputs, train_labels
```

```python
models,predicty,testy,r,p,heatmap,weights=Auto_Unet_Based_Model_pytorch(vy,vx,4,2,timestep=6,unet_mode='stt',test_size=0.2,valid_size=0.1,k_fold=None,task_mode='regression',if_best_mode='no',modelpath=None,ifrandom_split='no',if_bottom_attention_only='yes',if_resnet='yes',layer_time=1,baselayer=16,cov_kernelsize=3,cov_strides=1,pool_method='maxpooling',pool_kernel_size=2,pool_strides=2,ifnormalization='no',normalization_method=None,trans_num=1,num_heads=1,dim_feedforward=128,trans_dropout_rate=0.0,activate='tanh',nlp_activate='relu',if_last_act='no',if_print_model='yes',loss_function='default',optimizer='Adam',metrics='Pearsonr',learning_rate=0.0001,epochs=5000,batch_size=500,if_early_stopping=100,ifheatmap='no',ifweight='no',ifmute='no',ifsave='yes',savepath=str(MODEL_DIR / 'ST_TransUNet_5_24year_changjiang'),device='gpu')
```

```python
testy=testy.transpose(0,2,3,1)
predicty=predicty.transpose(0,2,3,1)
predicty=predicty*vy_std+vy_mean
testy=testy*vy_std+vy_mean
```
