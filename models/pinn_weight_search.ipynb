```python
import numpy as np
vx=np.zeros((3,u850.shape[0]-1,u850.shape[1],u850.shape[2]),dtype='float32')
vy=np.zeros((9,u850.shape[0]-1,u850.shape[1],u850.shape[2]),dtype='float32')
vx[0,:,:]=sp[:-1,:,:]
vx[1,:,:]=st[:-1,:,:]
vx[2,:,:]=sd[:-1,:,:]
vy[0,:,:,:]=u850[1:,:,:]
vy[1,:,:,:]=u925[1:,:,:]
vy[2,:,:,:]=u1000[1:,:,:]
vy[3,:,:,:]=v850[1:,:,:]
vy[4,:,:,:]=v925[1:,:,:]
vy[5,:,:,:]=v1000[1:,:,:]
vy[6,:,:,:]=w850[1:,:,:]
vy[7,:,:,:]=w925[1:,:,:]
vy[8,:,:,:]=w1000[1:,:,:]
vx=vx.transpose(1,2,3,0)
vy=vy.transpose(1,2,3,0)
print(vx.shape,vy.shape)
window_offset = 5
test_size, valid_size = 0.2, 0.1
test_start = int((1 - test_size) * len(vy))
train_stop = int((1 - valid_size / (1 - test_size)) * test_start)
train_inputs = vx[:train_stop + window_offset]
train_labels = vy[window_offset:train_stop + window_offset]
vx_mean = np.nanmean(train_inputs, axis=0)
vx_std = np.nanstd(train_inputs, axis=0)
vy_mean = np.nanmean(train_labels, axis=0)
vy_std = np.nanstd(train_labels, axis=0)
vx = (vx - vx_mean) / vx_std
vy = (vy - vy_mean) / vy_std
del train_inputs, train_labels
```

```python
import gc
import os
import hashlib
import numpy as np
import pandas as pd
import torch
from tqdm import tqdm


def _clear_cuda_memory():
    """Release the CUDA cache between trials."""
    gc.collect()
    if torch.cuda.is_available():
        try:
            with torch.cuda.device(torch.cuda.current_device()):
                torch.cuda.synchronize()
                torch.cuda.empty_cache()
                try:
                    torch.cuda.ipc_collect()
                except Exception:
                    pass
        except Exception:
            torch.cuda.empty_cache()
    gc.collect()


def _atomic_write_csv(df,path):
    if path is None:
        return
    directory=os.path.dirname(path)
    if directory:
        os.makedirs(directory,exist_ok=True)
    tmp=path+'.tmp'
    df.to_csv(tmp,index=False,encoding='utf-8-sig')
    os.replace(tmp,path)


def _checkpoint_key(lambda_cont,lambda_vor,lambda_shear):
    text=(
        f'{float(lambda_cont):.14e}|'
        f'{float(lambda_vor):.14e}|'
        f'{float(lambda_shear):.14e}'
    )
    return hashlib.sha1(text.encode('utf-8')).hexdigest()[:16]


def _sqlite_storage_url(db_path):
    absolute=os.path.abspath(db_path).replace('\\','/')
    return 'sqlite:///'+absolute


def Auto_tune_PINN_weights_1hour(
    vy,
    vx,
    vy_mean,
    vy_std,
    pinn_lat,
    pinn_lon,
    pinn_pressure_levels=(850,925,1000),
    pinn_w_external_scale=-0.1,

    n_trials=12,
    physics_group_weight=0.01,
    tune_global_weight=False,
    global_weight_bounds=(1e-4,1.0),
    ratio_bounds=(1e-2,1e2),
    ratio_base=(1.0,1.0,1.0),

    tuning_epochs=5000,
    tuning_patience=100,
    batch_size=500,
    learning_rate=1e-4,

    seed=25,

    result_csv=str(RESULT_DIR / 'PINN_weight_tuning_1hour.csv'),

    study_db=str(RESULT_DIR / 'PINN_weight_tuning_1hour_optuna.db'),
    study_name='DA_TransUNet_PINN_1hour',

    checkpoint_dir=str(MODEL_DIR / 'PINN_weight_tuning_1hour_checkpoints'),
    checkpoint_every=5,

    resume=True,
    use_optuna=True,
):
    """
    仅针对 1-h DA-TransUNet 自动调优三个 PINN loss 权重。

    ratio_base 用于指定三个物理项的基准相对比例。
    ratio_bounds 控制围绕该基准比例的乘法微调范围。
    当 ratio_base=(1,1,1) 时，与原始第一阶段逻辑完全一致。

    ============================================================
    真正的两级断点续传
    ============================================================
    1. trial 级：
       Optuna study 保存到 SQLite。
       已 COMPLETE 的 trial 重启后不会重新训练。

    2. 当前模型 epoch 级：
       每 checkpoint_every 个 epoch 保存：
         model_state_dict
         optimizer_state_dict
         epoch
         early-stopping 状态
       如果程序/Notebook 中断，重启后同一组参数会读取 checkpoint，
       从最近保存的 epoch 继续。

    ============================================================
    进度条
    ============================================================
    仅保留两个 tqdm：
      外层：PINN 参数 trial；
      内层：当前 trial 的 epoch。
    所有 train/val/ES 信息都通过 set_postfix 覆盖更新，
    不逐 epoch print 新行。

    ============================================================
    自动选择 objective
    ============================================================
    只使用 validation data MSE（标准化空间）选参数。
    held-out test set 不参与调参。
    """

    if not torch.cuda.is_available():
        raise RuntimeError('未检测到 CUDA。')
    cuda_device = torch.device(os.environ.get('YANGTZE_DEVICE', 'cuda:0'))
    if cuda_device.type != 'cuda':
        raise ValueError('This experiment requires a CUDA device.')
    torch.cuda.set_device(cuda_device.index if cuda_device.index is not None else 0)

    n_trials=int(n_trials)
    if n_trials<1:
        raise ValueError('n_trials 必须 >= 1。')

    physics_group_weight=float(physics_group_weight)
    if physics_group_weight<=0:
        raise ValueError('physics_group_weight 必须 > 0。')

    ratio_low,ratio_high=map(float,ratio_bounds)
    if ratio_low<=0 or ratio_high<=ratio_low:
        raise ValueError('ratio_bounds 必须满足 0 < low < high。')

    ratio_base_array=np.asarray(ratio_base,dtype=np.float64).reshape(-1)
    if ratio_base_array.size!=3:
        raise ValueError('ratio_base 必须包含 continuity、vorticity、shear 三个正数。')
    if (
        (not np.all(np.isfinite(ratio_base_array)))
        or np.any(ratio_base_array<=0)
    ):
        raise ValueError('ratio_base 中的三个值必须均为有限正数。')
    ratio_base_array=ratio_base_array/np.mean(ratio_base_array)

    gw_low,gw_high=map(float,global_weight_bounds)
    if gw_low<=0 or gw_high<=gw_low:
        raise ValueError(
            'global_weight_bounds 必须满足 0 < low < high。'
        )

    os.makedirs(checkpoint_dir,exist_ok=True)
    if study_db is not None:
        db_dir=os.path.dirname(study_db)
        if db_dir:
            os.makedirs(db_dir,exist_ok=True)

    pinn_pressure_levels=np.asarray(
        pinn_pressure_levels,
        dtype=np.float32
    ).reshape(-1)

    existing_records=[]
    if resume and result_csv is not None and os.path.exists(result_csv):
        try:
            old_df=pd.read_csv(result_csv)
            if 'validation_data_mse' in old_df.columns:
                old_df=old_df[
                    np.isfinite(
                        pd.to_numeric(
                            old_df['validation_data_mse'],
                            errors='coerce'
                        )
                    )
                ].copy()
            existing_records=old_df.to_dict('records')
        except Exception:
            existing_records=[]

    optuna=None
    study=None

    if use_optuna:
        try:
            import optuna
            optuna.logging.set_verbosity(optuna.logging.ERROR)

            sampler=optuna.samplers.TPESampler(
                seed=seed,
                n_startup_trials=min(5,max(1,n_trials//2))
            )

            storage=(
                _sqlite_storage_url(study_db)
                if study_db is not None
                else None
            )

            study=optuna.create_study(
                study_name=study_name,
                direction='minimize',
                sampler=sampler,
                storage=storage,
                load_if_exists=True,
            )

            if resume:
                stale=[]
                for frozen in list(study.trials):
                    if frozen.state==optuna.trial.TrialState.RUNNING:
                        stale.append(frozen)

                for frozen in stale:
                    params=dict(frozen.params)
                    try:
                        study.tell(
                            frozen.number,
                            state=optuna.trial.TrialState.FAIL
                        )
                    except Exception:
                        pass

                    required={
                        'ratio_continuity',
                        'ratio_vorticity',
                        'ratio_shear',
                    }
                    if required.issubset(params.keys()):
                        study.enqueue_trial(params)

            if len(study.trials)==0 and len(existing_records)==0:
                baseline={
                    'ratio_continuity':1.0,
                    'ratio_vorticity':1.0,
                    'ratio_shear':1.0,
                }
                if tune_global_weight:
                    baseline['global_physics_weight']=physics_group_weight
                study.enqueue_trial(baseline)

        except Exception:
            optuna=None
            study=None

    if study is not None:
        known_numbers={
            int(x['optuna_trial_number'])
            for x in existing_records
            if (
                'optuna_trial_number' in x
                and pd.notna(x['optuna_trial_number'])
            )
        }

        for frozen in study.trials:
            if frozen.state!=optuna.trial.TrialState.COMPLETE:
                continue
            if frozen.number in known_numbers:
                continue
            rec=dict(frozen.user_attrs)
            if 'validation_data_mse' not in rec:
                continue
            rec['optuna_trial_number']=int(frozen.number)
            existing_records.append(rec)

    successful_done=len(existing_records)

    if successful_done>=n_trials:
        results_df=pd.DataFrame(existing_records)
        results_df=results_df.sort_values(
            'validation_data_mse',
            ascending=True
        ).reset_index(drop=True)

        best_row=results_df.iloc[0]
        best_params={
            'lambda_continuity':
                float(best_row['lambda_continuity']),
            'lambda_vorticity':
                float(best_row['lambda_vorticity']),
            'lambda_shear':
                float(best_row['lambda_shear']),
            'validation_data_mse':
                float(best_row['validation_data_mse']),
            'physics_to_data_ratio':
                float(best_row.get('physics_to_data_ratio',np.nan)),
        }
        return best_params,results_df

    global_success_index=successful_done

    outer_bar=tqdm(
        total=n_trials,
        initial=successful_done,
        desc='外部 PINN 权重选择',
        position=0,
        leave=True,
        dynamic_ncols=True,
        mininterval=0.8,
        maxinterval=5.0,
        smoothing=0.05,
    )

    max_attempts=max(n_trials*3,n_trials+10)
    attempt_count=0

    try:
        while successful_done<n_trials:
            attempt_count+=1
            if attempt_count>max_attempts:
                raise RuntimeError(
                    '连续失败 trial 过多，已停止调优。'
                    '请检查最后一个模型报错。'
                )

            _clear_cuda_memory()

            np.random.seed(seed)
            torch.manual_seed(seed)
            torch.cuda.manual_seed_all(seed)

            model=None
            trial=None
            checkpoint_path=None

            try:
                if study is not None:
                    trial=study.ask()

                    ratio_cont=trial.suggest_float(
                        'ratio_continuity',
                        ratio_low,
                        ratio_high,
                        log=True
                    )
                    ratio_vor=trial.suggest_float(
                        'ratio_vorticity',
                        ratio_low,
                        ratio_high,
                        log=True
                    )
                    ratio_shear=trial.suggest_float(
                        'ratio_shear',
                        ratio_low,
                        ratio_high,
                        log=True
                    )

                    if tune_global_weight:
                        group_weight=trial.suggest_float(
                            'global_physics_weight',
                            gw_low,
                            gw_high,
                            log=True
                        )
                    else:
                        group_weight=physics_group_weight

                else:
                    if global_success_index==0:
                        ratio_cont=ratio_vor=ratio_shear=1.0
                        group_weight=physics_group_weight
                    else:
                        rng=np.random.default_rng(
                            seed+global_success_index
                        )
                        log_low=np.log(ratio_low)
                        log_high=np.log(ratio_high)
                        ratio_cont=float(
                            np.exp(rng.uniform(log_low,log_high))
                        )
                        ratio_vor=float(
                            np.exp(rng.uniform(log_low,log_high))
                        )
                        ratio_shear=float(
                            np.exp(rng.uniform(log_low,log_high))
                        )
                        if tune_global_weight:
                            group_weight=float(
                                np.exp(
                                    rng.uniform(
                                        np.log(gw_low),
                                        np.log(gw_high)
                                    )
                                )
                            )
                        else:
                            group_weight=physics_group_weight

                ratios=(
                    ratio_base_array
                    * np.asarray(
                        [ratio_cont,ratio_vor,ratio_shear],
                        dtype=np.float64
                    )
                )
                ratios=ratios/np.mean(ratios)

                lambda_cont=float(group_weight*ratios[0])
                lambda_vor=float(group_weight*ratios[1])
                lambda_shear=float(group_weight*ratios[2])

                key=_checkpoint_key(
                    lambda_cont,
                    lambda_vor,
                    lambda_shear
                )
                checkpoint_path=os.path.join(
                    checkpoint_dir,
                    f'pinn_trial_{key}.pth'
                )

                outer_bar.set_postfix(
                    {
                        'lc':f'{lambda_cont:.2e}',
                        'lv':f'{lambda_vor:.2e}',
                        'ls':f'{lambda_shear:.2e}',
                    },
                    refresh=False
                )

                model,val_score,val_components=Auto_DA_TransUnet_pytorch(
                    vy,
                    vx,
                    6,
                    4,
                    2,
                    test_size=0.2,
                    valid_size=0.1,
                    k_fold=None,
                    task_mode='regression',
                    if_best_mode='no',
                    modelpath=None,
                    ifrandom_split='no',
                    if_bottom_attention_only='yes',
                    if_resnet='yes',
                    layer_time=1,
                    baselayer=16,
                    cov_kernelsize=3,
                    cov_strides=1,
                    pool_method='maxpooling',
                    pool_kernel_size=2,
                    pool_strides=2,
                    ifnormalization='no',
                    normalization_method=None,
                    da_attention_order='space_first',
                    trans_num=1,
                    num_heads=1,
                    dim_feedforward=128,
                    trans_dropout_rate=0.0,
                    activate='tanh',
                    nlp_activate='relu',
                    if_last_act='no',

                    pinn_vy_mean=vy_mean,
                    pinn_vy_std=vy_std,
                    pinn_lambda_continuity=lambda_cont,
                    pinn_lambda_vorticity=lambda_vor,
                    pinn_lambda_shear=lambda_shear,
                    pinn_lat=pinn_lat,
                    pinn_lon=pinn_lon,
                    pinn_pressure_levels=pinn_pressure_levels,
                    pinn_w_external_scale=pinn_w_external_scale,

                    pinn_tuning_mode='yes',
                    pinn_inner_tqdm='yes',
                    pinn_inner_desc=(
                        f'内部模型 {successful_done+1}/{n_trials}'
                    ),

                    pinn_resume_checkpoint=(
                        'yes' if resume else 'no'
                    ),
                    pinn_checkpoint_path=checkpoint_path,
                    pinn_checkpoint_every=checkpoint_every,

                    if_print_model='no',
                    loss_function='default',
                    optimizer='Adam',
                    metrics='Pearsonr',
                    learning_rate=learning_rate,
                    epochs=tuning_epochs,
                    batch_size=batch_size,
                    if_early_stopping=tuning_patience,
                    ifheatmap='no',
                    ifweight='no',
                    ifmute='yes',
                    ifsave='no',
                    savepath=None,
                    device='gpu',
                )

                weighted_cont=(
                    lambda_cont
                    * float(val_components['continuity_loss'])
                )
                weighted_vor=(
                    lambda_vor
                    * float(val_components['vorticity_loss'])
                )
                weighted_shear=(
                    lambda_shear
                    * float(val_components['vertical_shear_loss'])
                )
                weighted_phys=(
                    weighted_cont
                    + weighted_vor
                    + weighted_shear
                )

                data_mse=float(val_components['data_loss'])
                physics_to_data=(
                    weighted_phys/data_mse
                    if data_mse>0
                    else np.nan
                )

                record={
                    'trial':successful_done+1,
                    'optuna_trial_number':(
                        int(trial.number)
                        if trial is not None
                        else np.nan
                    ),
                    'lambda_continuity':lambda_cont,
                    'lambda_vorticity':lambda_vor,
                    'lambda_shear':lambda_shear,
                    'global_physics_weight':group_weight,
                    'effective_ratio_continuity':float(ratios[0]),
                    'effective_ratio_vorticity':float(ratios[1]),
                    'effective_ratio_shear':float(ratios[2]),
                    'validation_data_mse':float(val_score),
                    'validation_continuity_loss':
                        float(val_components['continuity_loss']),
                    'validation_vorticity_loss':
                        float(val_components['vorticity_loss']),
                    'validation_vertical_shear_loss':
                        float(val_components['vertical_shear_loss']),
                    'weighted_continuity':weighted_cont,
                    'weighted_vorticity':weighted_vor,
                    'weighted_vertical_shear':weighted_shear,
                    'weighted_physics_total':weighted_phys,
                    'physics_to_data_ratio':physics_to_data,
                }

                if study is not None and trial is not None:
                    for k,v in record.items():
                        if isinstance(v,(np.floating,np.integer)):
                            v=v.item()
                        trial.set_user_attr(k,v)
                    study.tell(trial,float(val_score))

                existing_records.append(record)
                successful_done+=1
                global_success_index+=1

                results_df=pd.DataFrame(existing_records)
                _atomic_write_csv(
                    results_df,
                    result_csv
                )

                if (
                    checkpoint_path is not None
                    and os.path.exists(checkpoint_path)
                ):
                    try:
                        os.remove(checkpoint_path)
                    except Exception:
                        pass

                outer_bar.update(1)
                outer_bar.set_postfix(
                    {
                        'best':f"{results_df['validation_data_mse'].min():.4g}",
                        'last':f'{float(val_score):.4g}',
                    },
                    refresh=True
                )

            except KeyboardInterrupt:
                if study is not None and trial is not None:
                    pass
                raise

            except Exception as e:
                if study is not None and trial is not None:
                    try:
                        study.tell(
                            trial,
                            state=optuna.trial.TrialState.FAIL
                        )
                    except Exception:
                        pass

                outer_bar.set_postfix_str(
                    '当前 trial 失败；已清显存并继续',
                    refresh=True
                )

            finally:
                try:
                    del model
                except Exception:
                    pass
                _clear_cuda_memory()

    finally:
        outer_bar.close()

    results_df=pd.DataFrame(existing_records)

    results_df=results_df[
        np.isfinite(
            pd.to_numeric(
                results_df['validation_data_mse'],
                errors='coerce'
            )
        )
    ].copy()

    results_df=results_df.sort_values(
        'validation_data_mse',
        ascending=True
    ).reset_index(drop=True)

    if len(results_df)==0:
        raise RuntimeError('没有任何成功的 PINN 调优 trial。')

    best_row=results_df.iloc[0]

    best_params={
        'lambda_continuity':
            float(best_row['lambda_continuity']),
        'lambda_vorticity':
            float(best_row['lambda_vorticity']),
        'lambda_shear':
            float(best_row['lambda_shear']),
        'validation_data_mse':
            float(best_row['validation_data_mse']),
        'physics_to_data_ratio':
            float(best_row.get('physics_to_data_ratio',np.nan)),
        'trial':
            int(best_row['trial']),
    }

    return best_params,results_df
```
